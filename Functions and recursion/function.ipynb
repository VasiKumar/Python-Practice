```python
def average(a, b,c):
    avg=(a+b+c)/3
    return avg
print(average(1,2,3))
```

```python
def length(list1):
    return len(list1)

def printing(list1):
    for item in list1:
        print(item, end=" ")
list2=[1,2,3,4,5]
print(length(list2))
printing(list2)
```

```python
def sum(a,b):
    return a+b

print(sum(3,8))
```

```python
def fact(n):
    if n==0 or n<0:
        return 1
   
    return n*fact(n-1)
print(fact(3))
```
